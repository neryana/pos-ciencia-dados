#### Modelagem dos Dados

**Tipo de Modelagem:** Estrela

**Tabela Fato das Instituições de Ensino Superior**

**Nome da Tabela:** default.fato_ies

**Dicionário dos Dados**

| nome da coluna | tipo | tamanho | descricao |
| -------- | ----- | ----- | ----------- |
| CO_IES | Num | 8 | Código único de identificação da IES |
| NO_IES | Char | 200 | Nome da IES
| SG_IES |Char | 20 | Sigla da IES
| SG_UF_IES |Char | 2 | Sigla da Unidade da Federação da sede administrativa ou reitoria da IES
| CO_MUNICIPIO | Num | 7 | Código do município da sede administrativa da IES
| NO_TIPO_REDE | Char | 7 | Rede de ensino


```python
%sql

-- CRIAR TABELA FATO  (FASE PRATA)

CREATE OR REPLACE TABLE default.fato_ies AS 
SELECT
  CO_IES,
  NO_IES,
  SG_IES,
  SG_UF_IES,
  CO_MUNICIPIO_IES AS CO_MUNICIPIO, 
CASE
  WHEN TP_REDE = 1 THEN 'Pública'
  WHEN TP_REDE = 2 THEN 'Privada'
END AS NO_TIPO_REDE
FROM default.microdados_ed_sup_ies_2023;

SELECT
  count(*)
FROM default.fato_ies;
```

**Tabela Dimensão das Regiões de Planejamento do Ceará**

**Nome da Tabela:** default.dim_regiao_planejamento_ce

**Dicionário de Dados**

| nome da coluna | tipo | tamanho | descricao |
| -------- | ----- | ----- | ----------- |
| CO_MUNICIPIO | Num| 7 | Código do município da sede administrativa da IES |
| NO_MUNICIPIO | Char | 150 | Nome do Município da sede administrativa ou reitoria da IES
| REG_PLA_CE | Char | 100 | Nome da região de planejamento do Ceará.




```python
%sql

-- CRIAR TABELA DIMENSÃO REGIÕES DE PLANEJAMENTO DO CEARÁ (FASE PRATA)

CREATE OR REPLACE TABLE default.dim_regiao_planejamento_ce AS 
SELECT
  reg.`Código do município (IBGE)` AS CO_MUNICIPIO,
  reg.`Nome do município` AS NO_MUNICIPIO,
  reg.`Região de Planejamento` AS REG_PLA_CE
FROM default.regioes_planejamento_ceara AS reg;

SELECT 
COUNT(*)
FROM default.dim_regiao_planejamento_ce;
```

**Tabela Dimensão dos Docentes**

**Nome da Tabela:** default.dim_docentes

**Dicionário de Dados**

| nome da coluna | tipo | tamanho | descricao |
| -------- | ----- | ----- | ----------- |
| CO_IES | Num | 8 | Código único de identificação da IES |
| QT_DOC_TOTAL | Num | 8 | Quantidade total de docentes (em exercício e afastados)
| QT_DOC_EXE | Num | 8 | Quantidade total de docentes em exercício
| QT_DOC_EX_DOUT | Num | 8 | Quantidade de docentes em exercício com doutorado
| QT_DOC_EX_PRETA | Num | 8 | Quantidade de docentes em exercício - Cor/Raça preta
| QT_DOC_EX_INDIGENA | Num | 8 | Quantidade de docentes em exercício - Cor/Raça indígena


```python
%sql

-- CRIAR TABELA DIMENSÃO DOS DOCENTES (FASE PRATA)

CREATE OR REPLACE TABLE default.dim_docentes AS 
SELECT 
    CO_IES,
    QT_DOC_TOTAL,
    QT_DOC_EXE,
    QT_DOC_EX_DOUT,
    QT_DOC_EX_PRETA,
    QT_DOC_EX_INDIGENA
FROM default.microdados_ed_sup_ies_2023;


SELECT
*
FROM default.dim_docentes;
```

**BLOCO 1: PANORAMA DE IES POR **MUNICÍPIO****

**1. Quais municípios do estado do Ceará têm IES?**

```python
%sql

SELECT 
  DISTINCT reg.NO_MUNICIPIO
FROM default.fato_ies ies
LEFT JOIN default.dim_regiao_planejamento_ce reg
  ON reg.CO_MUNICIPIO = ies.CO_MUNICIPIO
WHERE reg.CO_MUNICIPIO IS NOT NULL
ORDER BY reg.NO_MUNICIPIO ASC;
```

**2. Quais municípios do estado do Ceará não têm IES?**

```python
%sql

SELECT 
  DISTINCT reg.NO_MUNICIPIO
FROM default.dim_regiao_planejamento_ce reg
LEFT JOIN default.fato_ies ies
  ON reg.CO_MUNICIPIO = ies.CO_MUNICIPIO
WHERE ies.CO_IES IS NULL
ORDER BY reg.NO_MUNICIPIO ASC;
```

**3. Quais municípios do estado do Ceará têm IES da rede pública?**

```python
%sql

SELECT 
  DISTINCT reg.NO_MUNICIPIO
FROM default.fato_ies ies
RIGHT JOIN default.dim_regiao_planejamento_ce reg
  ON reg.CO_MUNICIPIO = ies.CO_MUNICIPIO
WHERE ies.NO_TIPO_REDE = 'Pública'
ORDER BY reg.NO_MUNICIPIO ASC;
```


**4. Quais municípios do estado do Ceará tem IES da rede privada?**


```python
%sql

SELECT 
  DISTINCT reg.NO_MUNICIPIO
FROM default.fato_ies ies
RIGHT JOIN default.dim_regiao_planejamento_ce reg
  ON reg.CO_MUNICIPIO = ies.CO_MUNICIPIO
WHERE ies.NO_TIPO_REDE = 'Privada'
ORDER BY reg.NO_MUNICIPIO ASC;
```

**BLOCO 3: PANORAMA DE DOCENTES - QUANTITATIVO GERAL E DOCENTES COM DOUTORADO**


### **Solução do Problema/Conclusão**


Considerando o problema pressuposto deste trabalho, o baixo quantitativo de sedes administrativas de instituições de ensino superior (IES) da rede pública no Ceará e a necessidade de aumentar o número de docentes com doutorado nessas sedes, observando dados que justifiquem a ampliação de políticas afirmativas para pessoas pretas e indígenas, a análise dos dados obtidos por meio de consulta SQL a partir de três tabelas criadas na modelagem estrela (Tabela FATO: Tabela_FATO_IES, Tabela Dimensão: Tabela_REGIAO_PLANEJAMENTO e Tabela Dimensão: Tabela_DOCENTES) evidenciou informações relevantes para o tratamento desse problema.

**No panorama de IES por município e região de planejamento**, a análise de dados revelou que dos 184 municípios do Ceará, somente 5 possuem IES da rede pública: Crato, Fortaleza, Juazeiro do Norte, Redenção e Sobral.

Apesar de existirem IES em todas as 14 regiões de planejamento do estado do Ceará, apenas em cinco delas existem IES da rede pública: Cariri, Grande Fortaleza, Maciço de Baturité e Sertão de Sobral. Isso corrobora com a necessidade de ampliação de IES públicas no estado, focando especialmente nas regiões sem nenhuma IES, maiores e mais distantes daquelas onde já existem essas instituições, tais como: Sertão Central, Sertão de Canindé, Sertão dos Crateús e Vale do Jaguaribe.

**No que concerne aos docentes com doutorado**, o  levantamento apontou que dos 7.075 docentes em exercício na rede pública, 4.588 possuem doutorado, ou seja, o quadro de doutores não chega a 70%. No entanto, na rede privada, esse valor é bem menor, com somente 2.216 profissionais com esse nível de formação. Isso evidencia que o nível de conhecimento do corpo docente das IES de rede pública é maior em termos quantitativos em relação aos docentes das IES de rede privada.

A análise de dados aponta também que o maior número de docentes doutores no Ceará está na região da Grande Fortaleza e o menor no Sertão de Sobral, indicativo importante para saber onde está a maior necessidade de ampliação desse nível de professores por região de planejamento.

Por fim, **no que se refere ao número de docentes declarados pretos na rede pública**, os dados apontaram uma queda brusca em relação ao quantitativo total de docentes, caindo de 7.075 (total) para 275 classificados nesse grupo, ou seja, valor inferior a 4% de representatividade, estando a maior parte de docentes declarados pretos concentrada na região da Grande Fortaleza (143) e a menor na região do Cariri (27).

**A representividade de docente declarada indígena**, por sua vez, é ainda menor, somente 30 docentes indígenas em um universo 7.075 professores, com a maior parte na Grande Fortaleza (18) e a menor no Maçiço de Baturiré (2).

**Em suma**, a análise dos dados apurados explicitam a necessidade de expansão de IES para outras regiões do estado, como Sertão Central, Sertão de Canindé, Sertão dos Crateús e Vale do Jaguaribe; bem como a necessidade de ampliar o número de doutores na região do Sertão de Sobral. Os resultados apontaram, ainda, a demanda de políticas afirmativas mais efetivas para ampliar o quantitativo de  docentes declarados preto e indígenas, como, por exemplo, um percentual maior de bolsas destinadas para esses grupos e a ampliação de cotas em concurso público com vistas a reparar essa diferença, avaliando a ampliação e redistribuição de vagas considerando também o quantitativo existente nas regiões de planejamentos do estado.
